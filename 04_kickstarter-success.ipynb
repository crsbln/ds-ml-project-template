# Kickstarter Project Success: EDA and Modelling

We use the Kickstarter projects dataset to predict whether a crowdfunding campaign will succeed, and to analyse what drives success on the platform.

## The Machine Learning Workflow

### Define the Business Goal

The stakeholder is a **potential project creator** weighing up how to fund a project. Kickstarter uses an all-or-nothing model: a project only receives money if it reaches its goal.

Predict whether a project will be **successful** from information known at launch (category, goal, country, campaign length), so creators can judge whether Kickstarter is a good funding option for them and what their chances are.

### Get the Data

**Source:** [Kickstarter Projects on Kaggle](https://www.kaggle.com/datasets/kemical/kickstarter-projects) (file `ks-projects-201801.csv`).

**Download steps** (once per teammate, the raw CSV is not on GitHub):

1. Download the dataset archive from the Kaggle page above (a free Kaggle account is needed).
2. Unzip it and copy `ks-projects-201801.csv` into the `data/` folder of this repository.
3. `.gitignore` excludes `data/ks-projects-*.csv`, so the file is never committed.

We load the CSV and parse `deadline` and `launched` as dates, so we can later calculate things like the campaign length.

```python
import pandas as pd

kickstarter = pd.read_csv("data/ks-projects-201801.csv", parse_dates=["deadline", "launched"])
```

The dataset has 378,661 projects described by 15 columns: text (name, category, currency, country, state), numbers (goal, pledged, backers and their USD conversions) and two dates (`launched`, `deadline`).

Missing values are rare: 4 projects have no `name`, and 3,797 have no `usd pledged`. `usd_pledged_real` and `usd_goal_real` are complete, so they are the better USD columns to use.

`state` is the target. Besides `failed` and `successful` it also contains `canceled`, `undefined`, `live` and `suspended`, which we need to decide how to handle before modelling.
